Problema:

magine que você tem um DataFrame de vendas. Com o Pandera, você define o contrato de qualidade assim:

```python
# Realizando os imports
!pip install pandera
import pandas as pd
import pandera as pa
from pandera import Column, Check
```

```python
# Definindo o esquema de validação
schema = pa.DataFrameSchema({
    "id": Column(int, unique=True),
    "produto": Column(str),
    "preco": Column(float, Check.greater_than(0)), # O preço deve ser maior que zero
    "data": Column(pa.DateTime)
})
```

```python
# Dados de exemplo
df = pd.DataFrame({
    "id": [1, 2],
    "produto": ["Teclado", "Mouse"],
    "preco": [150.0, -10.0], # O -10.0 vai causar erro!
    "data": pd.to_datetime(["2023-01-01", "2023-01-02"])
})
```

```python
# Validando
schema.validate(df)
```
